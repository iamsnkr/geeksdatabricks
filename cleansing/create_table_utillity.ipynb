##### get schema details and create table based on the schema

```python
%sql
CREATE DATABASE IF NOT EXISTS geeks_az_dbs.cleansed_geekcoders
```

```python
def pre_schema(location):
    try:
        df=spark.read.format('delta').load(f'{location}').limit(1)
        schema=""
        for i in df.dtypes:
           schema=schema+i[0]+" "+i[1]+","
        return schema[0:-1]
    except Exception as err:
        print("Error Ocurred ",str(err))
```

```python
def f_delta_cleansed_load(table_name, location, database):
    catalog = "geeks_az_dbs"
    try:
        schema = pre_schema(location)
        spark.sql(f"DROP TABLE IF EXISTS {catalog}.{database}.{table_name}")
        spark.sql(f"""
            CREATE TABLE IF NOT EXISTS {catalog}.{database}.{table_name}
            ({schema})
            USING delta
            LOCATION '{location}'
        """)
    except Exception as err:
        print("Error Occurred:", str(err))
```
