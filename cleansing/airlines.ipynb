#### Reading airlines data using autoloader

```python
from pyspark.sql.functions import to_date, current_timestamp

# Reading data
df=spark.read.json("/mnt/geeks/rw_adls/airlines/*")

df = df.withColumn("Date_Part", to_date(current_timestamp()))

display(df)
```

```python
from pyspark.sql.functions import explode

df1=df.select(explode("response"),"Date_Part")
df_final=df1.select("col.*","Date_Part")

display(df_final)
```

```python
# Writing data
df_final.write.format("delta").mode("overwrite").save("/mnt/geeks/cld_adls/airlines")
```
