```python
%sql

USE DATABASE geek_marts;
```

```python
%sql
CREATE TABLE IF NOT EXISTS Reporting_Flight (
  date date,
  ArrDelay int,
  DepDelay int,
  Origin string,
  Cancelled string,
  CancellationCode string,
  UniqueCarrier string,
  FlightNum int,
  TailNum string,
  deptime string,
  date_year int
) USING DELTA LOCATION 'abfss://marts@geekadlsstoragesink.dfs.core.windows.net/reporting_flight'
```

```python
%sql
 INSERT
   OVERWRITE Reporting_Flight
 SELECT
   date,
   ArrDelay,
   DepDelay,
   Origin,
   Cancelled,
   CancellationCode,
   UniqueCarrier,
   FlightNum,
   TailNum,
   deptime,
   year(date) as date_year
 FROM
   cleansed_geekcoders.flight
```
