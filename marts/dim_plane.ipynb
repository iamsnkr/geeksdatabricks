```python
%sql
CREATE TABLE IF NOT EXISTS Dim_Plane (
  tailid STRING,
  type STRING,
  manufacturer STRING,
  issue_date Date,
  model STRING,
  status String,
  aircraft_type String,
  engine_type String,
  year int
) USING DELTA LOCATION 'abfss://marts@geekadlsstoragesink.dfs.core.windows.net/plane'
```

```python
%sql
INSERT OVERWRITE Dim_Plane
  SELECT
    tailid,
    type,
    manufacturer,
    issue_date,
    model,
    status,
    aircraft_type,
    engine_type,
    year
  FROM
    cleansed_geekcoders.plane
```
