```python
from datetime import date
import tabula
```

```python
def f_source_pdf_datalake(source_path, sink_path, output_format, page, file_name):
    try:
        path = dbutils.fs.mkdirs(f"{sink_path}{file_name.split('.')[0]}/date_{date.today()}/")
        tabula.convert_into(
            f"{source_path}{file_name}",
            f"{sink_path}{file_name.split('.')[0]}/date_{date.today()}/{file_name.split('.')[0]}.{output_format}",
            output_format=output_format,
            pages=page
        )
    except Exception as e:
        print(f"Error Occurred \n{e}")
```
