```python

tables = [
    "dim_airlines",
    "dim_airport",
    "dim_cancellation",
    "dim_plane",
    "dim_uniquecarrier",
    "reporting_flight"
]


def create_dbtables(tables_list):
  for table in tables_list:
    table_name = table.split("_")[1]
    delta_table = f"geeks_az_dbs.geek_marts.{table}"
    jdbc_url = "jdbc:sqlserver://geeksserver.database.windows.net:1433;database=geeeksdb"
    connection_properties = {
       "user": "geeksadmin@geeksserver",
       "password": "Test@123",
       "driver": "com.microsoft.sqlserver.jdbc.SQLServerDriver"
    }

    df = spark.read.format("delta").table(delta_table)

    df.write \
    .jdbc(url=jdbc_url, table=f"dbo.golden_{table_name}", mode="overwrite", properties=connection_properties)

    df.show(5, truncate=False)
  
    print(f"Data copied from Delta table {table} to Azure SQL Database successfully.")


```

```python
create_dbtables(tables)
```
